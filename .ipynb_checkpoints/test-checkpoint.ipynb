```python
import pandas as pd
import glob
```

```python
datasets = ['data_1', 'data_2']
files = ['train.bed', 'val.bed', 'test.bed']

for dataset in datasets:
    df_list = [pd.read_csv(f'data/test_data/{dataset}/{file}', sep='\t', header=None) for file in files]
    merged_df = pd.concat(df_list, ignore_index=True)

    merged_df.columns = ['chr', 'start', 'end'] + list(merged_df.columns[3:])
    
    valid_chr = ['chr' + str(i) for i in range(1, 23)] + ['chrX', 'chrY']
    filtered_df = merged_df[merged_df['chr'].isin(valid_chr)]
    
    dmr = filtered_df.groupby('chr').agg({'start': 'min', 'end': 'max'}).reset_index()
    
    dmr.to_csv(f'data/test_data/{dataset}/dmr.bed', sep='\t', index=False, header=None)
```

```python
meta = pd.read_csv('config/meta.tsv', sep='\t')
```

```python
meta['dmr'] = meta['train']
```

```python
meta['dmr'] = meta['dmr'].apply(lambda x:x.replace('train', 'dmr'))
```

```python
meta = meta[['name', 'train', 'validation', 'test', 'dmr', 'target', 'background']]
```

```python
meta.to_csv('config/meta.tsv', sep='\t', index=False)
```
